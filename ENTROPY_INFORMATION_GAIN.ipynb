```python
from __future__ import division
from math import log
```

```python
def entropy(pi):
  total = 0
  for p in pi:
    p = p / sum(pi)
    if p!= 0:
      total += p * log(p, 2)
    else:
        total +=0
  total *= -1
  return total
```

```python
def gain(d, a):
    total = 0
    for v in a:
        total+= sum(v) / sum(d)* entropy(v)

    gain = entropy(d) - total
    return gain
```

```python
willwait = [6, 6]
patron = [[4,0],[2,4],[0,2]]
print(gain(willwait, patron))
```

```python
playTennies = [9, 5]
outlook = [
    [4, 0],[2, 3],[3, 2]
]
temperature = [[2, 2],[3, 1],[4, 2]]
humidity = [[3, 4],[6, 1]]
wind = [[6, 2],[3,3]]
print(gain(playTennies, outlook))
print(gain(playTennies, temperature))
print(gain(playTennies, humidity))
print(gain(playTennies, wind))
```
